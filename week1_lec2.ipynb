```python
from peft import get_peft_model, LoraConfig
from transformers import AutoModelForCausalLM, AutoTokenizer

MODEL_NAME = "Qwen/Qwen2.5-1.5B"
print(f"Using model: {MODEL_NAME}")
```

```python
# Load the tokenizer of an open model (downloads config on first run)
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

text = "Hom nay troi dep"
# Encode text into a list of integer token IDs
token_ids = tokenizer.encode(text)
print(token_ids)                       # e.g. [39, 296, 299, 323, ...]
# Decode back to see the sub-word pieces
print(tokenizer.convert_ids_to_tokens(token_ids))
```

# 2.3

```python
model = AutoModelForCausalLM.from_pretrained(MODEL_NAME, device_map="auto", trust_remote_code=True)
inputs = tokenizer("Today weather is", return_tensors="pt").to(model.device)
```

```python
outputs = model.generate(**inputs, max_new_tokens=50)
print(tokenizer.decode(outputs[0], skip_special_tokens=True))
```
